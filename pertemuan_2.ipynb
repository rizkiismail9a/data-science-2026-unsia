```python
import pandas as pd

titanicDf = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

# isnull membuat DataFrame baru dengan nilai True di mana ada nilai yang hilang, dan False jika tidak ada.
missing_values = titanicDf.isnull().sum() # return Series
# Mengapa operasi Series dibagi skalar dapat berjalan?
# Ketika Anda membagi Pandas Series dengan skalar (integer len(titanicDf)),
# Pandas akan secara otomatis menerapkan operasi pembagian tersebut ke setiap elemen dalam Series.
# Jadi, setiap jumlah missing value di setiap kolom (yang merupakan nilai dalam Series) akan dibagi dengan total jumlah baris.
missing_percentage = (titanicDf.isnull().sum() / len(titanicDf)) * 100

missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percentage
})

# Filter to show only columns with missing values
missing_info = missing_info[missing_info['Missing Count'] > 0]

print(missing_info.sort_values(by='Missing Percentage', ascending=False))
print(missing_info.shape)
```

```python
import pandas as pd

titanicDf = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

classOneFemale = titanicDf[(titanicDf['Pclass'] == 1) & (titanicDf['Sex'] == 'female')]
average_age_class_one_female = classOneFemale['Age'].mean()

print(f'Jumlah penumpang wanita kelas 1: {len(classOneFemale)}')
# .2f Meminta agar angka ditampilkan dengan dua digit desimal setelah koma.
print(f'Rata-rata usia penumpang wanita kelas 1: {average_age_class_one_female:.2f} tahun')
```

```python
import pandas as pd

titanicDf = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

avgSurvivedByClass = titanicDf.groupby('Pclass')['Survived'].mean()
print(avgSurvivedByClass)


aggregationSurvivedByClass = titanicDf.groupby('Pclass').agg(
total = ('PassengerId', 'count'),
selamat = ('Survived', 'sum'),
pct_selamat= ('Survived', 'mean'),
rata_usia = ('Age', 'mean'),
max_fare = ('Fare', 'max')
)

print(aggregationSurvivedByClass.round(2))
```
