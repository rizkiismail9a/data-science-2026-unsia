```python
# Matplotlib: Grafik Dasar

import matplotlib.pyplot as plt
import numpy as np

# Cara 1:  Satu plot (figure dan satu axes)
fig, ax = plt.subplots(figsize=(8, 5))

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Label di sini adalah label di legenda
ax.plot(x, y, color="#028090", linewidth=2, label='sin(x)')

# Judul grafik
ax.set_title('Grafik Fungsi Sinus', fontsize=4)

# Label sumbu x
ax.set_xlabel('x (radian)')

# Label sumbu y
ax.set_ylabel('sin(x)')
ax.legend()
ax.grid(True, alpha=0.3)

# Ini untuk mencegah overlapping jika ada dua grafik
plt.tight_layout()
plt.show()
```

```python
# Matplotlib: Grafik Dasar

import matplotlib.pyplot as plt
import numpy as np

# Cara 2: Menampilkan beberapa grafik sekaligus
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(x, np.sin(x), color='#36ADA3')
axes[0].set_title('Grafik Sinus')

axes[1].plot(x, np.cos(x), color='#F68048')
axes[1].set_title('Grafik Cosinus')

plt.suptitle('Grafik Fungsi Trigonometri', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Bar Chart: Membandingkan Antar Kategori

import matplotlib.pyplot as plt
import numpy as np

kategori = ['Makanan', 'Fashion', 'Elektronik', 'Kesehatan', 'Hiburan']
val_2024 = [42, 31, 28, 19, 14]
val_2025 = [48, 27, 35, 25, 18]

# np.arange(5) (where np is an alias for numpy) generates an array of evenly spaced values starting from 0 up to (but not including) 5. So, it will produce [0, 1, 2, 3, 4].
x = np.arange(len(kategori))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))

# Bar chart grouped
bars1 = ax.barh(x - width/2, val_2024, width, label='2024', color='#028090', edgecolor='white')
bars2 = ax.barh(x + width/2, val_2025, width, label='2025', color='#F68048', edgecolor='white')

ax.set_yticks(x)
ax.set_yticklabels(kategori)
ax.set_xlabel('Pengeluaran (juta rupiah)')
ax.set_title('Distribusi Pengeluaran Konsumen per Kategori')
ax.legend()

# Menambahkan label angka di ujung setiap bar
ax.bar_label(bars1, padding=3, fontsize=9)
ax.bar_label(bars2, padding=3, fontsize=9)
# ax.spines[['top', 'right']].set_visible(False) # Hapus border atas dan kanan
plt.tight_layout()
plt.savefig('distribusi pengeluaran konsumen.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Scatter Plot dan Pair Plot
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('iris')

# Scatter plot dengan hue, size, dan garis regresi
fig, ax = plt.subplots(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species',
    size='petal_width',
    sizes=(30, 200),
    palette='Set2',
    alpha=0.7,
    ax=ax
)

# Garis regresi keseluruhan
sns.regplot(data=df,
x='sepal_length', y='petal_length',
scatter=False, color='gray',
line_kws={'linestyle':'--'}, ax=ax)
ax.set_title('Sepal Length vs Petal Length')
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot dan Pair Plot
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('iris')

# ── Pair plot: eksplorasi semua kombinasi variabel ──
# fig, ax = plt.subplots(figsize=(9, 6))

pair = sns.pairplot(df,
hue='species',
palette='Set2',
diag_kind='kde', # diagonal: KDE
plot_kws={'alpha': 0.6})
pair.fig.suptitle('Pair Plot — Iris Dataset', y=1.02)
plt.show()
```

```python
# Latihan

import seaborn as sns
import matplotlib.pyplot as plt, matplotlib.gridspec as gridspec
import pandas as pd, numpy as np

# Load Dataset
df = sns.load_dataset('penguins')

print(df.shape, df.dtypes)
print(df.describe().round(2))

fig, ax = plt.subplots(figsize=(16, 12))
fig.suptitle('Dashboard Analisis Pinguin', fontsize=18, fontweight='bold', y=0.98)

# Grid 2 x 2 subplot
gs = gridspec.GridSpec(2, 2, figure=fig,
hspace=0.4, wspace=0.35)
ax1 = fig.add_subplot(gs[0, 0]) # kiri atas
ax2 = fig.add_subplot(gs[0, 1]) # kanan atas
ax3 = fig.add_subplot(gs[1, 0]) # kiri bawah
ax4 = fig.add_subplot(gs[1, 1]) # kanan bawah

# Buat bar chart
rata2_massa = df.groupby('species')['body_mass_g'].mean()
rata2_massa.plot(kind='bar', ax=ax1, color='#028090', edgecolor='white')
ax1.set_title('Rata-rata Massa Pinguin Berdasarkan Jenis')
ax1.set_ylabel('Rata-rata Massa (gram)')
ax1.set_xlabel('Jenis Pinguin', fontweight='bold')
ax1.tick_params(axis='x', rotation=0)
ax1.spines[['top', 'right']].set_visible(False)

# Histogram
sns.histplot(data=df, x='bill_length_mm', hue='sex', kde=True, palette='Set2', ax=ax2)
ax2.set_title('Distribusi Panjang Paruh Pinguin Berdasarkan Jenis Kelamin')
ax2.set_xlabel('Panjang Paruh (mm)')
ax2.set_ylabel('Jumlah Pinguin')
ax2.spines[['top', 'right']].set_visible(False)

# Boxplot
sns.boxplot(data=df, x='species', y='body_mass_g', ax=ax3, palette='Set2', hue='sex')
ax3.set_title('Perbandingan Massa Pinguin Berdasarkan Jenis')
ax3.set_xlabel('Jenis Pinguin')
ax3.set_ylabel('Massa Pinguin')
ax3.spines[['top', 'right']].set_visible(False)

# Scatter Plot
sns.scatterplot(data=df, x='bill_length_mm', y='bill_depth_mm', hue='species', palette='Set2', ax=ax4, alpha=0.7)
sns.regplot(data=df, x='bill_length_mm', y='bill_depth_mm', scatter=False, color='gray', ax=ax4)
ax4.set_title('Perbandingan Panjang dan Lebar Paruh Pinguin Berdasarkan Jenis')
ax4.spines[['top', 'right']].set_visible(False)

# Tampilkan
plt.show()



```
